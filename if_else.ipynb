Else part is the default plan that one if part or elif is executed the else part will be executed.

```python
if x!=z:
    print('equal')
elif x==y:
    print('not equal')
else:
    print('else part')
```

```python
print("if part") if y>x else print("else part")
```

```python
print('if part') if x > y else print('else part')
```

```python
#Nested if - 
if x!=y:
    print('equal')
    if x < y:
        print('else part')
```

```python
# Difference between Nested if and if elif -
# Nested if - if the outer if condition is true then only inner if condition will be checked.
# if elif - if the first if condition is false then the flow of the program will go to elif condition and checked that condition.
```
