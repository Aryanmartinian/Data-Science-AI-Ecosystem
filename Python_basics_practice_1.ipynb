```python
# Python Program 1 - 
def calculation(a,b):
    product = a*b
    if product <= 1000:
        return product
    else:
        return a+b
    
num1 = int(input("Enter the first number:"))
num2 = int(input("Enter the second number:"))
result = calculation(num1,num2)
print("The result is:", result)
```

```python
# Python Program 2 -  Cumulative Sum of a Range 
def cumulative_sum(n):
    total = 0
    for i in range(0,n):
        total += i 
        print("Current Number",i,"Previous Number",i-1,"Sum:",total)
    
num = int(input("Enter the range"))
cumulative_sum(num)
```

```python
# Python Program 3 - String Indexing and Even Slicing - 
def even_char(s):
    for i in s[::2]: #  String slicing 
        print(i)

string = input("Enter a string:")
even_char(string)
```

```python
# Python Program 4 - String Slicing and Substring removal - 
def remove_substring(word,n):
    print('Original string:', word)
    res = word[n:] # starting index = n and ending index = end of the string
    return res

print("Removing characters from a string")
string = input("Enter a string:")
print(remove_substring(string,4))
print(remove_substring(string,2))

```

```python
# Python Program 5 - Variable Swapping (The In-place Method) - 
def swap(a,b):
    print("Before swapping:",a,b)
    a,b = b,a # in-place swapping
    print("After swapping:",a,b)

a = int(input("Enter the first character:"))
b = int(input("Enter the second character:"))
swap(a,b)
```

```python
# Python Program 6 - Calculating Factorial with a Loop - 
def factorial(n):
    fact = 1
    for i in range(1,n+1):
        fact *=i
    return fact # return statement should be placed outside the loop
num = int(input("Enter a number:"))
print(factorial(num))
```

```python
# Python Program 8 - String Reversal -
def reverse_string(s):
    reversed = s[::-1] # string slicing to reverse the string
    return reversed

s = input("Enter the string")
print("Reverse String:", reverse_string(s))
```

```python
# Python Program 8.2 - 
def reverse_str(s):
    reversed = ""
    for char in s[-1::-1]: # reverse order
        reversed += char
    return reversed

s = input("Enter the string")
print("Reverse string:", reverse_str(s))
```

```python
# Python Program 9 - Vowel Frequency Counter - 
sentence = input("Enter a sentence:")
vowels = "aeiou"
count = 0
for char in sentence.lower():
    if char in vowels:
        count +=1
print("Number of vowels:", count)
```

```python
# Python Program 10 - Finding Extremes in a List
nums = [45,2,89,12,7]
largest = max(nums) # in-built function
smallest = min(nums) # in-built function
print("Largest number:",largest)
print("Smallest number:",smallest)
```
