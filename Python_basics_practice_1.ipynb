```python
# Python Program 2 -  Cumulative Sum of a Range 
def cumulative_sum(n):
    total = 0
    for i in range(0,n):
        total += i 
        print("Current Number",i,"Previous Number",i-1,"Sum:",total)
    
num = int(input("Enter the range"))
cumulative_sum(num)
```

```python
# Python Program 4 - String Slicing and Substring removal - 
def remove_substring(word,n):
    print('Original string:', word)
    res = word[n:] # starting index = n and ending index = end of the string
    return res

print("Removing characters from a string")
string = input("Enter a string:")
print(remove_substring(string,4))
print(remove_substring(string,2))

```

```python
# Python Program 5 - Variable Swapping (The In-place Method) - 
def swap(a,b):
    print("Before swapping:",a,b)
    a,b = b,a # in-place swapping
    print("After swapping:",a,b)

a = int(input("Enter the first character:"))
b = int(input("Enter the second character:"))
swap(a,b)
```

```python
# Python Program 6 - Calculating Factorial with a Loop - 
def factorial(n):
    fact = 1
    for i in range(1,n+1):
        fact *=i
    return fact # return statement should be placed outside the loop
num = int(input("Enter a number:"))
print(factorial(num))
```

```python
# Python Program 8 - String Reversal -
def reverse_string(s):
    reversed = s[::-1] # string slicing to reverse the string
    return reversed

s = input("Enter the string")
print("Reverse String:", reverse_string(s))
```

```python
# Python Program 8.2 - 
def reverse_str(s):
    reversed = ""
    for char in s[-1::-1]: # reverse order
        reversed += char
    return reversed

s = input("Enter the string")
print("Reverse string:", reverse_str(s))
```

```python
# Python Program 16 - Palindrome logic on Numbers - 
def check_palindrome(number):
    print("original number", number)
    original_str = str(number)
    reversed_str = original_str[::-1] # string reverse logic keep this in mind
    if original_str == reversed_str:
        print("Yes, given number is palindrome number")
    else:
        print("No, given number is not palindrome number")

check_palindrome(121)
check_palindrome(125)
```

```python
# Python Program 17 - Merging Lists with Parity Filtering -
# odd numbers from list1 and even numbers from list2
# list1 = [10,20,25,30,35]
# list2 = [40,45,60,75,90]

def merge_list(list1,list2):
    result_list = []
    # Get odd numbers from list1 
    for num in list1:
        if num % 2!=0:
            result_list.append(num)
    # Get even numbers from list2
    for num in list2:
        if num % 2 == 0:
            result_list.append(num)
    return result_list

list1 = [10,20,25,30,35]
list2 = [40,45,60,75,90]
print("Merged List:", merge_list(list1,list2))
```

```python
# Python Program 18 - Integer Digit Extraction and Reversal -
# number = 7536 
# O/p - 6 3 5 7
def digit_extraction(num):
    print("Original number:", num)
    while num > 0:
        digit = num % 10 # get the last digit 
        print(digit, end=" ")
        num = num // 10 # remove the last digit

number = int(input("Enter a number:"))
digit_extraction(number)
```

```python
# Python Program 22 - Custom Exponentiation Function - 
def exponent(base,exp):
    num = exp
    result = 1
    # Repeat multiplication until num becomes 0
    while num > 0:
        result = result * base
        num = num - 1
    print(base, "raises to the power of", exp, "is:",result)

exponent(2,5)
exponent(5,4)
```
