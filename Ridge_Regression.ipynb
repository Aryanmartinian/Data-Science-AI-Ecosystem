```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import make_regression
x,y = make_regression(n_features=1,noise=5, n_samples=10000)
plt.scatter(x,y,s=2)
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y, test_size=0.2, random_state=0)
```

```python
from sklearn.linear_model import Ridge
reg = Ridge() # No alpha value is given
reg.fit(x_train,y_train)
```

```python
reg.predict(x_test)
```

```python
reg.score(x_test,y_test)
```

```python
reg.coef_ # The feature coefficient is very much high so this will occur high computational cost.
```

```python
reg.intercept_
```

```python
y_pred = reg.predict(x)
```

```python
plt.scatter(x,y,label='Actual',s=2)
plt.scatter(x,y_pred,label='Predicted',s=2)
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.show()
```

```python
from sklearn.linear_model import Ridge
reg = Ridge(alpha=100)
reg.fit(x_train,y_train)
reg.score(x_test,y_test)
```

```python
y_pred = reg.predict(x_test)
```

```python
y_pred2 = reg.predict(x)
```

```python
plt.scatter(x,y,label='Actual',s=2)
plt.scatter(x,y_pred2,label='Predicted',s=2)
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.show()
```

```python
reg.coef_
```

#### Changing the value of alpha and noticing the changes in the magnitude of the coefficient values -

```python
reg = Ridge(alpha=100000)
reg.fit(x_train,y_train)
reg.score(x_test,y_test)
```

```python
y_pred = reg.predict(x)
```

```python
plt.scatter(x,y,label='Actual',s=2)
plt.scatter(x,y_pred,label='Predicted',s=2)
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.show()
```

```python
reg.coef_
```

```python
### Again changing the value of alpha and notice the observations -
reg = Ridge(alpha=10000000)
reg.fit(x_train,y_train)
reg.score(x_test,y_test)
```

```python
y_pred = reg.predict(x)
plt.scatter(x,y,label='Actual',s=2)
plt.scatter(x,y_pred,label='Predicted',s=2)
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.show()
```

```python
reg.coef_
```

Imp Points 
- The coefficient exactly cannot be zero even how big alpha value will be.
- The alpha value will increase and coefficient value will be decrease but accuracy score will also be get affected and due to this the accuracy score will be decrease.
