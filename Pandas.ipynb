```python
# Pandas - used to handle the data
# Pandas used to create the data frames and series 
# To simply install pandas library in local system use command pip install pandas
import numpy as np
import pandas as pd
```

```python
s1 = pd.Series(data=[10,20,30,40,50])
s1
```

In Pandas the data created in two forms -
1. Series - It is one simple column of the table 
2. Dataframe - widely used method 

```python
s2 = pd.Series(data=[100,200,300,400,500])
s2
```

```python
s3 = pd.Series(data=[1000,2000,3000,4000,5000],index = list('vwxyz'))
s3
```

```python
# Creating DataFrame - A table like structure which has rows and columns like CSV(Comma Separated Values)
df = pd.DataFrame({'C1':s1,'C2':s2})
df
```

```python
df = pd.DataFrame(data=np.random.randint(1,100,size=(1000,5)),columns=list("abcde"))
df 
```
