```python
# Append method is depricated in pandas library so we can use concat method to append the data frames 
```

```python
# Merge Two DataFrames - 
data1 = pd.DataFrame(data=np.random.randint(1,100,size=(100,3)),columns=list('abc'))
```

```python
data2 = pd.DataFrame(data=np.random.randint(1,100,size=(100,3)),columns=list('ayz'))
```

```python
data1.merge(data2,on='a') # one column will be common in both data frames 
```

The merge method is used to merge two data frames based on a common column. In this case, the common column is 'a'. The resulting data frame will contain all the columns from both data frames, and the rows will be matched based on the values in the 'a' column.

The difference between merge method and append method is that in append or concat method there is no need to have a common column between two dataframes as it append it at the end of data1 dataframe but in merge there is should be a common column as it will be merge it like it is shown below - 
11 22 33 --> data 1 - These are the values of dataframe 1 a = 11, b=22,c=33
11 45 67 --> data 2- These are the values of dataframe 2 a = 11, y=45, z=67
Now the merge method will find the common value of a and merge it into one row -
11 22 33 45 67 --> a = 11 b =22 c = 33 y = 45 z = 67
