- one-dimensional array is the collection of 0th dimensional array --> Eg - [5,2,1]
- two-dimensional array is the collection of one dimensional array --> Eg - [[1,2,3],[4,5,6]]
- three-dimensional array is the collection of two-dimensional array --> Eg - [[[2,3,1],[4,5,6]],[[7,8,9],[10,11,12]]]
- zero-dimensional array is the collection of 1 element --> Eg - 5

```python
import numpy as np
a0 = np.array(10)
```

```python
print(a0)
```

```python
a0.ndim # ndim shows the dimension of the array. Here it is 0 because it is a scalar value.
```
