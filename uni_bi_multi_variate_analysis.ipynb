```python
import pandas as pd
```

```python
from sklearn.datasets import load_iris
iris = load_iris()
iris_df = pd.DataFrame(iris.data, columns=iris.feature_names)
```

```python
iris_df
```

```python
iris_df['target'] = iris.target # Add target Variable to the dataframe and name the variable - target
iris_df.head()
```

```python
iris_df.shape # 150 rows and 5 columns (4 features and 1 target variable)
```

### Uni-Variate Analysis

```python
# univariate analysis means take one variable and perform the analysis on that variable.
setosa = iris_df[iris_df['target'] == 0]
```

```python
setosa # Containing the samples that belong to class 0
```

```python
versicolor = iris_df[iris_df['target'] == 1]
```

```python
vergnica = iris_df[iris_df['target'] == 2]
```

```python
versicolor.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

In the below matplotlib plot the np.zeros_like function is producing the exact number of zeros for all the values that are present in that dataframe.
For Eg - If in setosa dataframe the samples of sepal length are 50 then the np.zeros_like function will produce 50 zeros values 

```python
plt.plot(setosa['sepal length (cm)'],np.zeros_like(setosa['sepal length (cm)']),'s') # on x-axis we have sepal length and on y-axis we have 0 for setosa class
plt.plot(versicolor['sepal length (cm)'],np.zeros_like(versicolor['sepal length (cm)']),'o') # on x-axis we have sepal length and on y-axis we have 0 for versicolor 
plt.plot(vergnica['sepal length (cm)'],np.zeros_like(vergnica['sepal length (cm)']),'d') 
```

- Conclusion - So according to the uni-variate analysis it is concluded that if sepal length is less than 4.8 cm then it is belong to setosa class by observing the graph. 
- Drawback - Univariate analysis cannot give complete information it provides less information because it is taking only one variable consideration.

### Bi-variate Analysis


In Bi-variate two variables are used sepal length and petal length 

```python
# Bi-variate Analysis can be achieved using scatter plot -
plt.scatter(iris_df['sepal length (cm)'],iris_df['petal length (cm)'],c=iris_df['target']) # x-axis = sepal length y-axis = petal length 
```
