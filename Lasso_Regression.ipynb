```python
from sklearn.datasets import make_regression
x,y = make_regression(n_features=1,n_samples=10000,random_state=0)
import matplotlib.pyplot as plt
plt.scatter(x,y,s=3)
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
from sklearn.linear_model import Lasso
reg = Lasso()
reg.fit(x_train,y_train)
```

```python
reg.predict(x_test)
```

```python
reg.score(x_test,y_test)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
y_pred = reg.predict(x)
```

```python
plt.scatter(x,y,label='Actual',s=3)
plt.scatter(x,y_pred,label='Predicted',s=3)
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.show()
```

### Now Changing the Alpha Value -

```python
reg = Lasso(alpha=2)
reg.fit(x_train,y_train)
reg.score(x_test,y_test)
```

```python
reg.coef_
```

```python
y_pred = reg.predict(x)
```

```python
plt.scatter(x,y,label='Actual',s=3)
plt.scatter(x,y_pred,label='Predicted',s=3)
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```

Imp Point - 
- If we increase the alpha value then the coefficient value will decrease plus the accuracy score or model score value will also decrease.
- If we decrease the alpha value then the coefficient value will increase plus the accuracy score or model score will also increase.
- To find the optimal value of Alpha we should use RidgeCV and LassoCV cross validation techniques.
