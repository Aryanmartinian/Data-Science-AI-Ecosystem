```python
import pandas as pd
df = pd.read_csv("age2.csv")
```

```python
df.head()
```

```python
import matplotlib.pyplot as plt
plt.scatter(df['Age'],df['Income'])
plt.xlabel('Age')
plt.ylabel('Income')
plt.show()
```

```python
plt.scatter(df['Degrees'],df['Income'])
plt.xlabel('Degrees')
plt.ylabel('Income')
plt.show()
```

```python
x = df[['Age','Degrees']]
y = df['Income']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(x,y,test_size=0.3,random_state=0)
```

```python
X_train
```

```python
X_test
```

```python
from sklearn.linear_model import LinearRegression
clf = LinearRegression()
```

```python
clf.fit(X_train,Y_train)
```

```python
clf.predict(X_test)
```

```python
clf.score(X_test,Y_test) # score methods internally use predict method and then compare the outputs with Y_test and give the accuracy
```

- Age = coefficient value = -20.370
- Degrees = -351.851

```python
clf.coef_ # The amount of coefficient weight of the features 
```

```python
clf.intercept_
```

- Why the coefficients are negative ? 
- Ans - Slope is negative (for eg - if age is increasing the income is decreasing) inversely proportional relation.

```python
y_pred = clf.predict(x)
y_pred
```
