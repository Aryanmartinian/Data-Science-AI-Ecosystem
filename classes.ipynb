```python
class class1:
    v1 = "Aryan"
    v2 = 1

obj1 = class1()
print(obj1)
print(obj1.v1)
print(obj1.v2)
```

```python
obj2 = class1()
print(obj2.v1)
obj2.v1 = "Satyaprem"
print(obj2.v1)
```

```python
class class2:
    name = "Aryan"
    roll_no = 1

    def abc(self): # Self is a reference to the current instance of the class. Here obj1 will be self passed to the function abc.
        print("name-->",self.name)

obj11 = class2()
obj11.abc()
obj11.name = "Satyaprem"
obj11.abc()
```

```python
class class3:
    def __init__(self,name):
        self.name = name
    roll_no = 1

    def abc(self):
        print("name-->",self.name)

obj111 = class3("Manisha") #  Here we are passing the value "Manisha" to the constructor __init__ which will initialize the name attribute of the class.
print(obj111.name)
obj111.abc()
```
