- Parent Node Entropy -

```python
entropy_results = -(5/10)*np.log2(5/10) - (5/10)*np.log2(5/10)
entropy_results
```

- Now let's calculate the entropy of each feature and check whose has more information gain. The feature which has more information gain the data will be split according to that feature.

```python
data.rename(columns = {'5me':'yt'}, inplace = True)
```

```python
data[data.yt == 1]
```

- Calculating the Entropy of Yt with yes samples -

```python
entropy_results_yt_yes = -(5/6)*np.log2(5/6) -(1/6)*np.log2(1/6)
entropy_results_yt_yes
```

```python
data[data.yt == 0]
```

- Calculating the Entropy of Yt feature with No samples -

```python
# Zero entropy because all the values are same.
```

- Calculating the Combined Entropy of Yt with Yes and No samples entropy - 

```python
yt_entropy = (6/10) * entropy_results_yt_yes +  (4/10) * 0
yt_entropy
```

```python
info_yt = entropy_results - (6/10) * entropy_results_yt_yes -  (4/10) * 0
info_yt
```

The Information Gain is 0.6099

- Calculating the Entropy of Guide feature with Yes samples -

```python
data[data.guide == 1]
```

```python
entropy_results_guide_yes = -(4/5)*np.log2(4/5) -(1/5)*np.log2(1/5)
entropy_results_guide_yes
```

- Calculating the Entropy of Guide feature with No samples -

```python
data[data.guide == 0]
```

```python
entropy_results_guide_no = -(1/5)*np.log2(1/5) -(4/5)*np.log2(4/5)
entropy_results_guide_no
```

- Combined Entropy of Guide feature with Yes and No data samples -

```python
guide_entropy = (5/10) * entropy_results_guide_yes +  (5/10) * entropy_results_guide_no
guide_entropy
```

```python
info_guide = entropy_results - guide_entropy
info_guide
```

- The information Gain from guide feature is 0.278
- The info gain(yt) > info gain(guide)
- Hence the splitting will happen according to yt feature

### Implementing the Decision Tree Classifier -

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
df = data
df.result = le.fit_transform(df.result)
df
```

```python
x = df[['yt','guide']]
y = df['result']
```

```python
from sklearn.tree import DecisionTreeClassifier, export_graphviz
```

```python
dt = DecisionTreeClassifier(criterion='entropy')
```

```python
dt.fit(x,y)
```

```python
export_graphviz(dt,'dt.tree',feature_names=['yt','guide'])
```

```python
dt.predict([[1,1]])
```

```python
dt.score(x,y)
```
