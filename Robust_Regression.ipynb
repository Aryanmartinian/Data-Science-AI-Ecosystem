```python
from sklearn.datasets import make_regression
import numpy as np
from sklearn.linear_model import LinearRegression,RANSACRegressor
import matplotlib.pyplot as plt
```

```python
x,y = make_regression(n_samples=200000,n_features=1,noise=10,random_state=0)
outliers = 30000
np.random.seed(2)
y[:outliers] = 600 + np.random.normal(size=(outliers))
```

```python
sample = np.random.normal(size=(outliers))
```

```python
len(sample)
```

```python
plt.scatter(x,y,s=3)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
lr = LinearRegression()
lr.fit(x_train,y_train)
```

```python
rr = RANSACRegressor()
rr.fit(x_train,y_train)
```

```python
rr_pred = rr.predict(x_test)
```

```python
lr_pred = lr.predict(x_test)
```

```python
plt.scatter(x_test,y_test,label='Actual data',c="green",s=2)
plt.scatter(x_test,rr_pred,label='RANSACRegressor',c="red")
plt.scatter(x_test,lr_pred,label='LinearRegression',c="black")
plt.legend()
```
