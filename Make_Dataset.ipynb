In this Make_Dataset file we will get the idea of how we can make our own dataset with the properties that we want to be present in the Dataset.

```python
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_regression
```

- n_features = no. of features = 1
- noise = To remove idealistic data pattern the noise parameter is used as the real world data
will be not so smooth and clear it has some dirty variations and noise.
- random_state = random_state parameter is used in the dataset when we don't want to change the combination of the data points in the datasets. If we don't use random_state parameter in the dataset creation time then whenever we run the below command the data points combination or values will changed so to make values of the dataset remain fixed for a specific model training or to increase the value of the accuracy of the model we use random_state.

```python
x,y = make_regression(n_features=1, noise=5, n_samples=10000,random_state=0) 
```

```python
x 
```

```python
y
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

Because of noise the above data has been spread out if noise was zero it was simple linear regression straight line.

### Creating Classification Dataset -

```python
from sklearn.datasets import make_classification
```

- n_features = no. of features present in the data = 5
- rows = no. of samples = 10000
- n_classes = no. of classes in the dataset = 2

```python
x,y = make_classification(n_classes=2, n_features=5,n_samples=10000)
plt.scatter(x[:,0],x[:,1],c=y)
```

### Creating Clustering(Unsupervised learning) Dataset - 

```python
from sklearn.datasets import make_blobs
```

- n_samples = no. of rows = 500
- centers = no. of blobs = 3
- n_features = no. of features = 2

```python
x,y = make_blobs(n_samples=500, centers=3, n_features=2, random_state=0)
```

```python
plt.scatter(x[:,0],x[:,1],c=y)
```
