```python
def function2(plus):
    print("Hello and Welcome to my world of",plus)

function2("Data Science")
```

```python
def name(fname,lname):
    print(fname+" "+lname)

name("Aryan","Mishra")
```

```python
name("Ashok","Samrat")
```

```python
def function2(*plus): # List parameter to pass multiple values in a function
    print("Hello and Welcome Dosto", plus[0],plus[1],plus[2],plus[3])

function2('to my world of','Data Science','and','Machine Learning.')
```

```python
# Dictionary parameter in the function - 
def name2(**name):
    print(name['fname']+" "+name['lname'])
# In the above function we are passing a dictionary as a parameter and we are accessing the values of the dictionary using the keys.
name2(fname="Aryan",lname="Mishra")
```

```python
def name3(**name): # Another example of dictionary parameter 
    print(name['s1'],name['s2'],name['s3'])

name3(s1="Aryan",s2="Mishra",s3="is a genius person")
```

```python
def function2(plus="Data World"): # Default parameter in the function
    print("Hello and Welcome to",plus)

function2('to dsa world')
function2()
```
