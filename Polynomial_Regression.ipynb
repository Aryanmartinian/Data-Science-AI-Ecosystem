```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([2,3,4,5,7,8,9])
y = np.array([4,6,8,5,4,7,8])

plt.scatter(x,y)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=0)
```

```python
x_train = x_train.reshape(-1,1) # Data was in horizontal form so converting it to vertical form
x_train
```

```python
x_test = x_test.reshape(-1,1)
x_test
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
model.fit(x_train,y_train)
```

```python
model.score(x_test,y_test)
```

```python
y_pred = model.predict(x_test)
y_pred
```

```python
# Due to splitting of dataset into training and testing we are not getting the same line as before and prediction is cannot done because the data points are very less.
x = x.reshape(-1,1)
```

```python
model = LinearRegression()
```

```python
model.fit(x,y)
```

```python
y_pred = model.predict(x_test)
plt.scatter(x,y)
plt.scatter(x_test,y_pred,color='red')  
```

```python
model.score(x,y)
```

### Now Implementing Polynomial Regression (Poly feature) --

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly = PolynomialFeatures(degree=2)
poly_x = poly.fit_transform(x.reshape(-1,1)) # independent variable data has been obtained
model1 = LinearRegression()
model1.fit(poly_x,y)
y_pred1 = model1.predict(poly_x)
plt.scatter(x,y) # Normal data points are present in blue color which default color of scatter plot.
plt.plot(x,y_pred1,color='red')
poly_x  
```

```python
model1.score(poly_x,y) # Score is also increased
```
