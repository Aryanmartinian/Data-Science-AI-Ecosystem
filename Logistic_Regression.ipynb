### Logistic Regression 
- Classification Model
- Works on Categorical Dataset (Target Variable has values 0 or 1)

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("job.csv")
```

```python
df.head(20)
```

```python
df.tail()
```

```python
# Splitting of the independent and dependent data variable(features) - 
x = df[['Years']]
y = df['result']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(x,y,train_size=0.8,random_state=1)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
```

```python
model.fit(X_train,Y_train)
```

```python
print(X_test)
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
Y_test
```

```python
model.score(X_test,Y_test)
```

```python
from sklearn.metrics import accuracy_score 
```

```python
accuracy_score(y_pred,Y_test) # This metric is generally used in classification model.
```

```python
import seaborn as sns
sns.regplot(x=x,y=y,logistic=True) # logistic = true will show how the s-curve of logistic regression will be displayed matlab s-curve kaisa dikh rha hai usske liye kara hai.
```
