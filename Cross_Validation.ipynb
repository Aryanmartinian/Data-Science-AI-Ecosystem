```python
import pandas as pd
from sklearn.datasets import load_iris
iris = load_iris()
table = pd.DataFrame(iris.data,columns=iris.feature_names)
table.head()
```

```python
iris
```

```python
table['target'] = iris.target
```

```python
table.head()
```

```python
x = table.drop(['target'],axis='columns')
y = iris.target
```

### Hold Out Method -

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(x,y,test_size = 0.20,random_state=0)
```

```python
from sklearn.naive_bayes import GaussianNB
classifier1 = GaussianNB()
classifier1.fit(X_train,Y_train)
```

```python
classifier1.score(X_test,Y_test)
```

### K-Fold Cross Validation -

```python
from sklearn.model_selection import cross_val_score,KFold # cross_val_score is used to check for each K Fold
```

```python
kf=KFold(n_splits=5)
kf
```

```python
for train, test in kf.split(x,y):
    print("TRAIN-->", train,'\n', "TEST-->", test,'\n')
```

```python
score=cross_val_score(classifier1,x,y,cv=kf)
```

```python
score # score of each iteration 1st iteration score --> 100 , 2nd iteration --> 0.9666667, 3rd iteration --> 0.9
```

```python
score.mean() # To measure kfold performance we take average value 
```

### Leave P Out Cross Validation -

- Very Exhaustive Cross Validation Method 

```python
from sklearn.model_selection import LeavePOut
```

```python
lp=LeavePOut(p=2)
lp
```

```python
lp.get_n_splits(x)
```

```python
for train, test in lp.split(x,y):
    print("Train-->", train,'\n','Test-->', test,'\n')
```

```python
classifier3 = GaussianNB()
```

```python
score = cross_val_score(classifier3,x,y,cv=lp)
```

```python
score
```

```python
score.mean()
```
